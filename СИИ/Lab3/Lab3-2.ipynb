Задание 1

```python
# === Бинарная классификация (LogisticRegression) — Colab-ноутбук, без argparse ===

import re
from pathlib import Path
from typing import Optional, Tuple
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, average_precision_score, confusion_matrix,
    RocCurveDisplay, PrecisionRecallDisplay, classification_report
)

# ---------- НАСТРОЙКИ ----------
CSV_PATH      = "/content/car_resale_prices.csv"
TARGET_COL    = None          # если есть готовая бинарная цель — укажи имя; иначе оставь None
PRICE_COL     = "resale_price" # будем делать цель: price > median
TEST_SIZE     = 0.2
RANDOM_STATE  = 42
MAX_ROWS      = 0
# --------------------------------

warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)

# ===== Вспомогательные =====
def find_binary_column(df: pd.DataFrame) -> Optional[str]:
    for c in df.columns:
        vals = df[c].dropna().unique()
        if len(vals) == 2:
            return c
    return None

def pick_numeric_price_like(df: pd.DataFrame) -> Optional[str]:
    price_like = [c for c in df.columns if re.search(r'price', c, re.IGNORECASE)]
    price_like = [c for c in price_like if pd.api.types.is_numeric_dtype(df[c])]
    if price_like:
        return price_like[0]
    numeric_cols = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]
    if numeric_cols:
        return df[numeric_cols].var(numeric_only=True).sort_values(ascending=False).index[0]
    return None

def drop_id_leakage_cols(X: pd.DataFrame, target_source_col: Optional[str]) -> pd.DataFrame:
    drop_like = set()
    for c in X.columns:
        if re.search(r'\b(id|vin|uuid|index|^Unnamed: ?0$)\b', str(c), re.IGNORECASE):
            drop_like.add(c)
    if target_source_col is not None and target_source_col in X.columns:
        drop_like.add(target_source_col)  # убираем исходную колонку таргета (чтобы не было лика)
    return X.drop(columns=list(drop_like)) if drop_like else X

def build_target(df: pd.DataFrame,
                 target_col: Optional[str],
                 price_col: Optional[str]) -> Tuple[pd.DataFrame, str, Optional[str]]:
    """Вернёт (df, '__target__', source_col). Если передан price_col — он ДОЛЖЕН быть числовым."""
    df = df.copy()
    # 1) Явно заданная цель
    if target_col is not None:
        if target_col not in df.columns:
            raise ValueError(f"--target-col '{target_col}' отсутствует.")
        s = df[target_col]
        if pd.api.types.is_numeric_dtype(s) and s.dropna().nunique() > 2:
            thr = s.median()
            df["__target__"] = (s > thr).astype(int)
            return df, "__target__", target_col
        vals = s.dropna().unique()
        if len(vals) == 2:
            mapping = {vals[0]: 0, vals[1]: 1}
            df["__target__"] = s.map(mapping).astype(int)
            return df, "__target__", target_col
        raise ValueError(f"'{target_col}' не бинарная и не числовая с >2 значениями.")
    # 2) Бинаризация по цене
    if price_col is not None:
        if price_col not in df.columns:
            raise ValueError(f"--price-col '{price_col}' отсутствует.")
        s = df[price_col]
        if not pd.api.types.is_numeric_dtype(s):
            raise ValueError(f"--price-col '{price_col}' должен быть числовым.")
        thr = s.median()
        df["__target__"] = (s > thr).astype(int)
        return df, "__target__", price_col
    # 3) Попытка найти готовую бинарную
    bin_col = find_binary_column(df)
    if bin_col is not None:
        s = df[bin_col]
        vals = s.dropna().unique()
        mapping = {vals[0]: 0, vals[1]: 1}
        df["__target__"] = s.map(mapping).astype(int)
        return df, "__target__", bin_col
    # 4) price-like
    price_like = pick_numeric_price_like(df)
    if price_like is None:
        raise ValueError("Не удалось определить целевую. Укажи TARGET_COL или PRICE_COL.")
    s = df[price_like]
    thr = s.median()
    df["__target__"] = (s > thr).astype(int)
    return df, "__target__", price_like

# ===== Загрузка =====
csv_path = Path(CSV_PATH)
if not csv_path.exists():
    raise FileNotFoundError(f"CSV не найден: {csv_path}")
df = pd.read_csv(csv_path)
if MAX_ROWS and len(df) > MAX_ROWS:
    df = df.head(MAX_ROWS).copy()

# ===== КОНВЕРТАЦИЯ resale_price -> число (рупии) =====
# Поддержка форматов: "₹ 5.45 Lakh", "₹ 7.62L", "1.2 Cr", "12,50,000", "50k", "Rs. 1,20,000", "1.1 million",
# а также диапазонов: "5.45 - 7.62 Lakh", "₹ 5.45 Lakh ₹ 7.62 Lakh" (берётся среднее).
if PRICE_COL is not None and PRICE_COL in df.columns and not pd.api.types.is_numeric_dtype(df[PRICE_COL]):
    def parse_indian_money(x):
        if pd.isna(x):
            return np.nan
        s = str(x).strip().lower()
        # убрать валюту/мусор
        s = (s.replace("₹", "").replace("rs.", "").replace("rs", "").replace("inr", "")
               .replace(",", "").replace("–", "-").replace("—", "-"))
        s = re.sub(r"\s+", " ", s)

        # множитель по единицам
        mult = 1
        if re.search(r"\b(cr|crore|crores)\b", s):
            mult = 10_000_000
        elif re.search(r"\b(l|lac|lakh|lakhs|lacs)\b", s):
            mult = 100_000
        elif re.search(r"\bm(illion)?\b", s):
            mult = 1_000_000
        elif re.search(r"\bk\b", s):
            mult = 1_000

        # извлечь числа (если два и более — берём среднее)
        nums = re.findall(r"\d+(?:\.\d+)?", s)
        if not nums:
            return np.nan
        val = float(np.mean([float(n) for n in nums]))
        return val * mult

    num_col = f"{PRICE_COL}_num"
    df[num_col] = df[PRICE_COL].apply(parse_indian_money)
    # удалить пустые конверсии
    before = len(df)
    df = df[df[num_col].notna()].reset_index(drop=True)
    after = len(df)
    print(f"[INFO] Конвертация '{PRICE_COL}' -> '{num_col}'. Удалено строк без цены: {before - after}")
    print(df[[PRICE_COL, num_col]].head(5))
    PRICE_COL = num_col  # дальше используем числовую колонку

# ===== Построение цели, препроцессинг, модель =====
df, target_name, source_col = build_target(df, TARGET_COL, PRICE_COL)
df = df.dropna(how='all').copy()

X = df.drop(columns=[target_name])
y = df[target_name].astype(int)
X = drop_id_leakage_cols(X, source_col)

numeric_features = [c for c in X.columns if pd.api.types.is_numeric_dtype(X[c])]
categorical_features = [c for c in X.columns if c not in numeric_features]

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler(with_mean=True)),
])
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore")),
])

preprocess = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features),
    ],
    remainder="drop"
)

log_reg = LogisticRegression(
    penalty="l2",
    C=1.0,
    solver="saga",            # быстрый для разреженных OHE
    class_weight="balanced",  # компенсирует дисбаланс классов
    max_iter=2000,
    tol=1e-4,
    random_state=RANDOM_STATE,
)

clf = Pipeline(steps=[("preprocess", preprocess),
                     ("logreg", log_reg)])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, stratify=y, random_state=RANDOM_STATE
)

clf.fit(X_train, y_train)

proba_test = clf.predict_proba(X_test)[:, 1]
y_pred = (proba_test >= 0.5).astype(int)

metrics = {
    "accuracy": accuracy_score(y_test, y_pred),
    "precision": precision_score(y_test, y_pred, zero_division=0),
    "recall": recall_score(y_test, y_pred, zero_division=0),
    "f1": f1_score(y_test, y_pred, zero_division=0),
    "roc_auc": roc_auc_score(y_test, proba_test),
    "pr_auc": average_precision_score(y_test, proba_test),
    "positive_rate_test": float(y_test.mean()),
    "positive_rate_pred": float(y_pred.mean()),
    "n_rows": int(len(df)),
    "n_train": int(len(X_train)),
    "n_test": int(len(X_test)),
    "n_num_features": len(numeric_features),
    "n_cat_features": len(categorical_features),
    "target_name": target_name,
    "target_source_col": source_col if source_col is not None else "",
}

print("\n=== Сводка данных ===")
print(f"Всего строк: {metrics['n_rows']}")
print(f"Train/Test: {metrics['n_train']} / {metrics['n_test']}")
print(f"Числовых признаков: {metrics['n_num_features']} | Категориальных: {metrics['n_cat_features']}")
print(f"Целевая: {metrics['target_name']} (источник: {metrics['target_source_col'] or '— авто'})")
print(f"Доля класса=1 в тесте: {metrics['positive_rate_test']:.3f}")

print("\n=== Метрики (тест) ===")
for k in ["accuracy", "precision", "recall", "f1", "roc_auc", "pr_auc"]:
    print(f"{k:>9}: {metrics[k]:.4f}")

cm = confusion_matrix(y_test, y_pred)
print("\nМатрица ошибок:")
print(pd.DataFrame(cm, index=["True 0", "True 1"], columns=["Pred 0", "Pred 1"]))

print("\nClassification report:")
print(classification_report(y_test, y_pred, digits=4))

out_dir = csv_path.parent
roc_path = out_dir / "roc_curve.png"
pr_path  = out_dir / "pr_curve.png"

plt.figure()
RocCurveDisplay.from_predictions(y_test, proba_test)
plt.title("ROC-кривая (Logistic Regression)")
plt.tight_layout()
plt.savefig(roc_path, dpi=150)
plt.show()

plt.figure()
PrecisionRecallDisplay.from_predictions(y_test, proba_test)
plt.title("PR-кривая (Logistic Regression)")
plt.tight_layout()
plt.savefig(pr_path, dpi=150)
plt.show()

print(f"\nФайлы сохранены рядом с CSV:\n  ROC: {roc_path}\n  PR : {pr_path}")
```

Выполните расчёты вручную для слайда 43 (на примере любых 3 промежуточных AUC ROC, поясните, как из промежуточных характеристик получаются 4 финальных). В рамках детализации ОЦИФРУЙТЕ ROC: в Excel, отрисуйте линиями в Word или отрисуйте отдельными отрезками Python-кодом (не используйте готовые функции для ROC). [баллов: 2]

```python
# === Multiclass AUC-ROC (ручной расчёт, без готовых ROC-функций) ===
# Данные как на слайде 43:
# Y — матрица истинных меток (4 объекта × 3 класса), S — матрица скорингов (та же размерность)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ---------- Определения (кратко) ----------
# macro: среднее AUC по классам поровну
# weighted: среднее AUC по классам с весами = поддержка класса (число положительных)
# micro: "сплющиваем" все классы в одну бинарную задачу и строим один ROC
# samples (пообъектный): AUC по каждому объекту сравнивая скора истинных vs ложных классов, затем среднее

# ---------- Данные со слайда ----------
Y = np.array([
    [1, 1, 0],  # x1
    [0, 1, 1],  # x2
    [0, 1, 0],  # x3
    [1, 0, 0],  # x4
], dtype=int)

S = np.array([
    [0.7, 0.6, 0.5],
    [0.3, 0.4, 0.6],
    [0.5, 0.9, 0.2],
    [0.4, 0.5, 0.1],
], dtype=float)

out_dir = Path(".")
out_dir.mkdir(parents=True, exist_ok=True)

# ---------- Вспомогательные функции (ручной ROC + AUC) ----------
def roc_points_auc(y_true, scores):
    """
    РУЧНОЙ расчёт ROC-точек и AUC (трапеции).
    - y_true: 0/1
    - scores: float
    Возвращает: (thresholds, points[(FPR,TPR)], auc)
    """
    y_true = np.asarray(y_true).astype(int)
    scores = np.asarray(scores).astype(float)
    P = (y_true == 1).sum()
    N = (y_true == 0).sum()

    uniq = np.unique(scores)[::-1]           # уникальные пороги по убыванию
    thresholds = np.r_[np.inf, uniq, -np.inf]

    pts = []
    for t in thresholds:
        y_pred = (scores >= t).astype(int)
        TP = ((y_pred == 1) & (y_true == 1)).sum()
        FP = ((y_pred == 1) & (y_true == 0)).sum()
        TPR = TP / P if P else 0.0
        FPR = FP / N if N else 0.0
        pts.append((FPR, TPR))

    # AUC (ручные трапеции)
    x = np.array([p[0] for p in pts])
    y = np.array([p[1] for p in pts])
    order = np.argsort(x, kind="mergesort")
    x = x[order]; y = y[order]
    auc = float(np.sum((x[1:] - x[:-1]) * (y[1:] + y[:-1]) / 2.0))
    return thresholds.tolist(), [(float(a), float(b)) for a,b in zip(x,y)], auc

def samples_auc_for_row(y_row, s_row):
    """
    Пообъектный AUC для одной строки (multi-label):
    вероятность, что скор правильного класса > скор неправильного (ничьи = 0.5).
    """
    pos_idx = np.where(y_row == 1)[0]
    neg_idx = np.where(y_row == 0)[0]
    if len(pos_idx) == 0 or len(neg_idx) == 0:
        return np.nan
    wins = 0.0
    total = 0
    for i in pos_idx:
        for j in neg_idx:
            total += 1
            if s_row[i] > s_row[j]:
                wins += 1.0
            elif s_row[i] == s_row[j]:
                wins += 0.5
    return wins / total if total else np.nan

def plot_roc(points, title, save_path):
    x = [p[0] for p in points]
    y = [p[1] for p in points]
    plt.figure()
    plt.plot(x, y, marker='o')            # кусочные линии + маркеры
    plt.plot([0, 1], [0, 1], linestyle='--')  # диагональ случайного
    plt.xlabel("FPR")
    plt.ylabel("TPR")
    plt.title(title)
    plt.xlim(0, 1)
    plt.ylim(0, 1)
    plt.grid(True, linestyle=':')
    plt.tight_layout()
    plt.savefig(save_path, dpi=150)
    plt.show()

# ---------- AUC по классам (one-vs-rest) ----------
K = Y.shape[1]
per_class = []
roc_csv_paths = []
for k in range(K):
    yk = Y[:, k]
    sk = S[:, k]
    thr, pts, auc = roc_points_auc(yk, sk)
    per_class.append({"class": k+1, "auc": auc, "support": int(yk.sum())})

    # Таблица ROC-точек -> CSV (для Excel)
    df_pts = pd.DataFrame(pts, columns=["FPR","TPR"])
    csv_path = out_dir / f"roc_points_class{k+1}.csv"
    df_pts.to_csv(csv_path, index=False)
    roc_csv_paths.append(str(csv_path))

    # График
    plot_roc(pts, f"ROC (class {k+1}) — AUC={auc:.3f}", out_dir / f"roc_class{k+1}.png")

per_class_df = pd.DataFrame(per_class)
print("Per-class AUC:")
print(per_class_df)

# ---------- Macro / Weighted ----------
macro_auc = per_class_df["auc"].mean()
weighted_auc = (per_class_df["auc"] * per_class_df["support"]).sum() / per_class_df["support"].sum()

# ---------- Micro: сплющиваем во все пары (все классы вместе) ----------
y_micro = Y.flatten()
s_micro = S.flatten()
_, micro_pts, micro_auc = roc_points_auc(y_micro, s_micro)
pd.DataFrame(micro_pts, columns=["FPR","TPR"]).to_csv(out_dir / "roc_points_micro.csv", index=False)
plot_roc(micro_pts, f"ROC (micro) — AUC={micro_auc:.3f}", out_dir / "roc_micro.png")

# ---------- Samples (пообъектный) ----------
sample_aucs = [samples_auc_for_row(Y[i], S[i]) for i in range(Y.shape[0])]
samples_auc = float(np.nanmean(sample_aucs))

print("\nFinal multi-class AUC aggregates:")
print(f"  Macro   : {macro_auc:.3f}")
print(f"  Weighted: {weighted_auc:.3f}")
print(f"  Micro   : {micro_auc:.3f}")
print(f"  Samples : {samples_auc:.3f}")

print("\nROC point tables saved for Excel:")
for p in roc_csv_paths + [str(out_dir / "roc_points_micro.csv")]:
    print(" -", p)

print("\nSample-wise AUC values:", [round(v,3) for v in sample_aucs])
```

Какой вариант «лучше» и когда
	•	Macro — когда важна равная важность всех классов (оценка справедливости/баланса качества).
	•	Weighted — когда важно отражать текущий дисбаланс классов (что чаще встретится, то сильнее влияет). Это самый «приземлённый» агрегат для продовых метрик.
	•	Micro — когда нужен общий «глобальный» скилл ранжирования по всем решениям сразу (часто завышен при доминирующих/лёгких классах).
	•	Samples — при multi-label и фокусе «качество для каждого объекта» (равный вес объектам, а не классам).

Почему AUC ROC падает при переходе к многоклассовым
	1.	One-vs-rest усложняет задачу: каждый класс отделяется от объединения остальных → больше «похожих» негативов, чаще путаница, ниже вероятность P(s^+ > s^-), значит ниже AUC.
	2.	Ранговые конфликты между классами: невозможно одним набором скорингов обеспечить, чтобы все истинные классы одновременно доминировали над всеми ложными — где-то ранжирование ломается.
	3.	Дисбаланс классов: при weighted/micro общий итог тянут вниз частые, но «тяжёлые» классы (у нас класс 2 с AUC = 0.667).
	4.	Больше негативов → труднее держать низкий FPR: при том же пороге сплющивание добавляет много «приграничных» пар, что ухудшает площадь под кривой.
	5.	Калибровка по классам различается: скоринги для разных классов не сопоставимы идеально; при усреднениях это проявляется как систематическое снижение.
